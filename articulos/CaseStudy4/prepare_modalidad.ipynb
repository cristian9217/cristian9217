**Importación de librerías**  

Se importan las librerías necesarias para análisis de datos y estadística:

```python
import pandas as pd 
```

**Carga del dataset**

Un archivo que contiene información y sus preferencias de estudio en las modalidades presencial, híbrido y a distancia.

```python
try:
  modalidad = pd.read_csv("modalidad_programas.csv")
  modalidad_copy = modalidad.copy()
  print("¡Archivo cargado con éxito y guardado como dataframe!")
  print(f"Cantidad de filas y columnas: {modalidad_copy.shape}")
except FileNotFoundError:
  print("Error: no se encontró el archivo modalidad_programas.csv.")
```

**Información del DataFrame `modalidad_copy`**

Muestra la estructura del DataFrame, incluyendo los nombres de las columnas, los tipos de datos y el conteo de valores no nulos.

```python
print("Información sobre el dataframe: ")
modalidad_copy.info()
```

**Eliminación de duplicados `df_modalidad`**

Elimina cualquier fila duplicada del conjunto de datos para garantizar la integridad de los datos.

```python
# Remove duplicates
df_modalidad = modalidad_copy.drop_duplicates().copy()

# Check for duplicates
duplicates = modalidad_copy[modalidad_copy.duplicated()]

if duplicates.empty:
    print("No se detectaron datos repetidos.")
else:
    print("Se encontraron datos repetidos:", duplicates)
```

**Transforma tus datos**

La transformación de datos es el proceso de convertir los datos de un formato a 
otro, preservando la limpieza e integridad del dataset. Este proceso hace que 
los datos sean más organizados, consistentes y útiles para el análisis.

```python
class DataCleaner:
    def __init__(self, df):
        self.df = df
    
    def get_df(self):
        """ Devuelve el DataFrame modificado."""
        return self.df
    
    def rename_columns(self, old_value, new_value):
        """ Renombra columnas de un dataframe. """
        col_index = self.df.columns.get_loc(old_value)
        self.df.columns.values[col_index] = new_value
    
    def update_modalidad(self, old_value, new_value):
        """ Actualiza los valores de la columna en el dataframe. """
        idx = self.df['Modalidad'] == old_value
        self.df.loc[idx, 'Modalidad'] = new_value
```

**Limpieza de Datos del Conjunto de Modalidad**  

Pasos realizados:  

1. Renombrar una columna del DataFrame.
2. Actualizar los valores de una modalidad a otra.  
3. Guardar el conjunto de datos limpio en un nuevo archivo CSV.  

```python
try:   
    data = DataCleaner(df_modalidad)

    data.rename_columns('Nivel Educativo', 'NivelEducativo')

    data.update_modalidad('Online', 'Distancia')

    df_modalidad = data.get_df()

except Exception as e:
    print(f"Error: {e}")
finally:
    df_modalidad.to_csv('cleaned_modalidad.csv', index=False)
    print("Los datos han sido limpiados y guardados en el archivo CSV.")
```
